```
public interface ITransactionLogger
{
    void LogTransaction(string message);
}

public interface IAuditService
{
    void AuditTransaction(decimal amount);
}

public class BankAccount 
{
    public string AccountNumber { get; private set; }
    public decimal Balance { get; protected set; }
    public string AccountType { get; protected set; }
    public int TransactionsCount { get; protected set; }
    public decimal DailyLimit { get; private set; }
    public DateTime CreationDate { get; private set; }
    public int ActiveDays { get; private set; }
    public decimal TotalTurnover { get; protected set; }

    public BankAccount(string accountNumber, decimal initialBalance, string accountType)
    {
        AccountNumber = accountNumber;
        Balance = initialBalance;
        AccountType = accountType;
        CreationDate = DateTime.Now;
        DailyLimit = 100000;
        ActiveDays = 0;
        TotalTurnover = 0;
        TransactionsCount = 0;
    }

    public virtual string GetInfo()
    {
        return $"Номер счета: {AccountNumber}, Баланс: {Balance}, Тип счета: {AccountType}";
    }

    public virtual void Deposit(decimal amount)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма должна быть положительной");
        
        Balance += amount;
        TotalTurnover += amount;
        TransactionsCount++;
        Console.WriteLine($"Новый баланс: {Balance} руб.");
    }

    public virtual void Withdraw(decimal amount)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма должна быть положительной");
            
        if (amount > Balance)
        {
            Console.WriteLine("Недостаточно средств");
            return;
        }

        Balance -= amount;
        TotalTurnover += amount;
        TransactionsCount++;
        Console.WriteLine($"Новый баланс: {Balance} руб.");
    }

    public void SetDailyLimit(decimal limit)
    {
        DailyLimit = limit;
        Console.WriteLine($"Установлен дневной лимит: {limit} руб.");
    }

    public string GetAccountStatus()
    {
        return $"Статус счета: {(Balance > 0 ? "Активный" : "Неактивный")}";
    }

    public decimal CalculateAverageTransaction()
    {
        return TransactionsCount > 0 ? TotalTurnover / TransactionsCount : 0;
    }

    public void UpdateActiveDays()
    {
        ActiveDays++;
        Console.WriteLine($"Ваш счет активен уже {ActiveDays} дней");
    }
}

public class SavingsAccount : BankAccount
{
    public decimal InterestRate { get; protected set; }
    public int DepositsCount { get; private set; }
    public decimal MaxBalance { get; private set; }
    public decimal MinBalanceRequirement { get; private set; }
    public int InterestAccrualPeriod { get; private set; }
    public decimal TotalInterestEarned { get; private set; }
    public DateTime LastInterestDate { get; private set; }

    public SavingsAccount(string accountNumber, decimal initialBalance, decimal interestRate) 
        : base(accountNumber, initialBalance, "Сберегательный")
    {
        InterestRate = interestRate;
        MinBalanceRequirement = 1000;
        InterestAccrualPeriod = 30;
        LastInterestDate = DateTime.Now;
        MaxBalance = 1000000;
        DepositsCount = 0;
    }

    public override void Deposit(decimal amount)
    {
        base.Deposit(amount);
        DepositsCount++;

        if (Balance >= MinBalanceRequirement)
        {
            decimal interest = amount * InterestRate / 100;
            TotalInterestEarned += interest;
            Balance += interest;
            LastInterestDate = DateTime.Now;
            Console.WriteLine($"Начислены проценты: {interest} руб.");
        }
    }

    public void SetMinBalanceRequirement(decimal amount)
    {
        MinBalanceRequirement = amount;
        Console.WriteLine($"Минимальное требование к балансу: {amount} руб.");
    }

    public void AccrueInterestManually()
    {
        if (Balance >= MinBalanceRequirement)
        {
            decimal interest = Balance * InterestRate / 100;
            TotalInterestEarned += interest;
            LastInterestDate = DateTime.Now;
            Balance += interest;
            Console.WriteLine($"Начислено процентов: {interest} руб.");
        }
    }

    public decimal GetProjectedInterest(int days)
    {
        return Balance * InterestRate / 100 * (days / (decimal)InterestAccrualPeriod);
    }

    public string GetInterestHistory()
    {
        return $"Заработано процентов: {TotalInterestEarned} руб.";
    }
}

public class CheckingAccount : BankAccount
{
    public decimal OverdraftLimit { get; private set; }
    public decimal MonthlyFee { get; private set; }
    public bool IsVip { get; private set; }
    public int FreeTransactionsLeft { get; private set; }
    public decimal TransactionCommission { get; private set; }
    public DateTime NextFeeDate { get; private set; }
    public int OverdraftUsageCount { get; private set; }
    public decimal UsedOverdraft { get; private set; }

    public CheckingAccount(string accountNumber, decimal initialBalance, decimal overdraftLimit) 
        : base(accountNumber, initialBalance, "Текущий")
    {
        OverdraftLimit = overdraftLimit;
        MonthlyFee = 50;
        FreeTransactionsLeft = 10;
        TransactionCommission = 10;
        NextFeeDate = DateTime.Now.AddMonths(1);
        IsVip = false;
        UsedOverdraft = 0;
    }

    public override void Withdraw(decimal amount)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма должна быть положительной");
        
        decimal availableAmount = Balance + OverdraftLimit;
        if (amount > availableAmount)
        {
            Console.WriteLine($"Недостаточно средств. Доступно: {availableAmount} руб.");
            return;
        }

        if (amount > Balance)
        {
            decimal overdraftUsed = amount - Balance;
            UsedOverdraft = overdraftUsed;
            OverdraftUsageCount++;
            Console.WriteLine($"Использован овердрафт на сумму {overdraftUsed} руб.");
        }
        
        Balance -= amount;
        TotalTurnover += amount;
        TransactionsCount++;

        if (FreeTransactionsLeft > 0)
            FreeTransactionsLeft--;
        else
        {
            Balance -= TransactionCommission;
            Console.WriteLine($"Списана комиссия за операцию: {TransactionCommission} руб.");
        }
            
        Console.WriteLine($"Новый баланс: {Balance} руб.");
        Console.WriteLine($"Использовано овердрафта: {Math.Max(0, -Balance)}/{OverdraftLimit} руб.");
    }

    public void SetTransactionCommission(decimal commission)
    {
        TransactionCommission = commission;
        Console.WriteLine($"Комиссия: {commission} руб.");
    }

    public decimal CalculateTotalMonthlyCosts()
    {
        return MonthlyFee + (TransactionsCount * TransactionCommission);
    }

    public string GetOverdraftStatus()
    {
        decimal usedOverdraft = Balance < 0 ? Math.Abs(Balance) : 0;
        string status = usedOverdraft > 0 ? "Используется" : "Не используется";
        return $"Статус овердрафта: {status} ({usedOverdraft}/{OverdraftLimit} руб.)";
    }

    public void UpdateFeeDate(DateTime newDate)
    {
        NextFeeDate = newDate;
        Console.WriteLine($"Обновлена дата следующей комиссии: {newDate:d}");
    }

    public void RepayOverdraft(decimal amount)
    {
        if (Balance >= 0)
        {
            Console.WriteLine("Овердрафт не используется");
            return;
        }

        decimal overdraftAmount = Math.Abs(Balance);
        decimal repayment = Math.Min(amount, overdraftAmount);
        
        Balance += repayment;
        Console.WriteLine($"Погашено овердрафта: {repayment} руб.");
        Console.WriteLine($"Остаток овердрафта: {Math.Max(0, -Balance)} руб.");
    }
}

public class PremiumAccount : BankAccount, ITransactionLogger, IAuditService
{
    public decimal InterestRate { get; protected set; }
    public decimal PremiumBonusRate { get; private set; }
    public int LoyaltyProgramPoints { get; private set; }
    public DateTime PremiumExpirationDate { get; private set; }
    public decimal TotalPremiumBenefits { get; private set; }
    public decimal TotalInterestEarned { get; private set; }
    public DateTime LastInterestDate { get; private set; }

    public PremiumAccount(string accountNumber, decimal initialBalance, decimal interestRate) 
        : base(accountNumber, initialBalance, "Премиум")
    {
        InterestRate = interestRate;
        PremiumBonusRate = 2.0m;
        LoyaltyProgramPoints = 1000;
        PremiumExpirationDate = DateTime.Now.AddYears(1);
        TotalPremiumBenefits = 0;
        TotalInterestEarned = 0;
        LastInterestDate = DateTime.Now;
    }

    public void LogTransaction(string message)
    {
        Console.WriteLine($"Банковская выписка счета {AccountNumber}: {message} - {DateTime.Now}");
    }

    public void AuditTransaction(decimal amount)
    {
        Console.WriteLine($"Аудиторская проверка счета {AccountNumber}: Сумма {amount}, Баланс {Balance}, Время {DateTime.Now}");
    }

    public override void Deposit(decimal amount)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма должна быть положительной");
        
        Balance += amount;
        TotalTurnover += amount;
        TransactionsCount++;
        decimal bonus = CalculatePremiumBonus(amount);
        Balance += bonus;
        TotalPremiumBenefits += bonus;
     
        AddLoyaltyPoints((int)(amount / 100));
        AccrueInterest();
        LogTransaction($"Депозит: {amount} руб; Полученный бонус: {bonus} руб.");
        AuditTransaction(amount);
        
        Console.WriteLine($"Новый баланс: {Balance} руб.");
    }

    public override void Withdraw(decimal amount)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма должна быть положительной");
            
        if (amount > Balance)
        {
            Console.WriteLine("Недостаточно средств");
            return;
        }

        Balance -= amount;
        TotalTurnover += amount;
        TransactionsCount++;
        
        LogTransaction($"Снятие: {amount} руб.");
        AuditTransaction(amount);
        
        Console.WriteLine($"Новый баланс: {Balance} руб.");
    }

    public void AddLoyaltyPoints(int points)
    {
        LoyaltyProgramPoints += points;
        Console.WriteLine($"Начислено баллов лояльности: {points}");
    }

    public decimal CalculatePremiumBonus(decimal amount)
    {
        return amount * PremiumBonusRate / 100;
    }

    private void AccrueInterest()
    {
        decimal interest = Balance * InterestRate / 100;
        TotalInterestEarned += interest;
        LastInterestDate = DateTime.Now;
        Balance += interest;
        Console.WriteLine($"Начислены премиум проценты: {interest} руб.");
    }

    public string GetPremiumStatus()
    {
        return $"Премиум статус действителен до: {PremiumExpirationDate:d}";
    }

    public string GetInterestHistory()
    {
        return $"Заработано процентов: {TotalInterestEarned} руб., Премиум бонусов: {TotalPremiumBenefits} руб.";
    }
}

SavingsAccount savings = new SavingsAccount("12345", 5000, 5);
Console.WriteLine(savings.GetInfo());
savings.Deposit(1000);
savings.AccrueInterestManually();

PremiumAccount premium = new PremiumAccount("99999", 10000, 5);
premium.Deposit(2000);
Console.WriteLine(premium.GetPremiumStatus());
Console.WriteLine(premium.GetInterestHistory());

CheckingAccount checking = new CheckingAccount("67890", 2000, 1000);
checking.Withdraw(2500);
Console.WriteLine(checking.GetOverdraftStatus());

premium.LogTransaction("Новая операция 001");
premium.AuditTransaction(1000);
```
