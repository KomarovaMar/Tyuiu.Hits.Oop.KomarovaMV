```
using System;
using System.Collections.Generic;
public delegate void AccountOperation(string message);
public delegate void BalanceChangedHandler(decimal newBalance, decimal oldBalance);

public interface ITransactionLogger
{
    void LogTransaction(string message);
}

public interface IAuditService
{
    void AuditTransaction(decimal amount);
}

public interface IAccountNotifier
{
    void SendNotification(string message);
}

public interface IBalanceCalculator
{
    decimal CalculateProjectedBalance(int months);
}

public class BankAccountCollection<T> where T : BankAccount
{
    private List<T> _accounts = new List<T>();

    public event AccountOperation OnCollectionChanged;

    public void Add(T account)
    {
        _accounts.Add(account);
        Console.WriteLine($"Счет {account.AccountNumber} добавлен в коллекцию");
        OnCollectionChanged?.Invoke($"Добавлен счет: {account.AccountNumber}");
    }

    public void Remove(T account)
    {
        _accounts.Remove(account);
        Console.WriteLine($"Счет {account.AccountNumber} удален из коллекции");
        OnCollectionChanged?.Invoke($"Удален счет: {account.AccountNumber}");
    }

    public void DisplayAllAccounts()
    {
        Console.WriteLine("\nВСЕ СЧЕТА В КОЛЛЕКЦИИ");
        foreach (var account in _accounts)
        {
            Console.WriteLine(account.GetInfo());
        }
    }

    public T FindAccount(string accountNumber)
    {
        return _accounts.Find(acc => acc.AccountNumber == accountNumber);
    }

    public int GetTotalAccounts()
    {
        return _accounts.Count;
    }

    public decimal GetTotalBalance()
    {
        decimal total = 0;
        foreach (var account in _accounts)
        {
            total += account.Balance;
        }
        return total;
    }
}

public class BankAccount 
{
    public string AccountNumber { get; private set; }
    public decimal Balance { get; protected set; }
    public string AccountType { get; protected set; }
    public int TransactionsCount { get; protected set; }
    public decimal DailyLimit { get; private set; }
    public DateTime CreationDate { get; private set; }
    public int ActiveDays { get; private set; }
    public decimal TotalTurnover { get; protected set; }
    public string OwnerName { get; set; }
    public string Currency { get; protected set; }
    public bool IsActive { get; protected set; }
    public decimal MinBalance { get; protected set; }
    public string BranchCode { get; set; }
    public DateTime LastActivityDate { get; protected set; }
    public bool IsOnlineBankingEnabled { get; set; }
    public string Email { get; set; }
    public string PhoneNumber { get; set; }
    public string Address { get; set; }

    public event BalanceChangedHandler OnBalanceChanged;

    public BankAccount(string accountNumber, decimal initialBalance, string accountType)
    {
        AccountNumber = accountNumber;
        Balance = initialBalance;
        AccountType = accountType;
        CreationDate = DateTime.Now;
        DailyLimit = 100000;
        ActiveDays = 0;
        TotalTurnover = 0;
        TransactionsCount = 0;
        OwnerName = "Неизвестный владелец";
        Currency = "RUB";
        IsActive = true;
        MinBalance = 0;
        Email = "не указан";
        PhoneNumber = "не указан";
        Address = "не указан";
    }

    public virtual string GetInfo()
    {
        return $"Счет: {AccountNumber}, Баланс: {Balance} {Currency}, Тип: {AccountType}, Владелец: {OwnerName}";
    }

    public virtual string GetInfo(bool detailed)
    {
        if (detailed)
        {
            return $"{GetInfo()}, Активных дней: {ActiveDays}, Транзакций: {TransactionsCount}";
        }
        return GetInfo();
    }

    public void DisplayContactInfo()
    {
        Console.WriteLine($"Контактная информация для счета {AccountNumber}:");
        Console.WriteLine($"Email: {Email}, Телефон: {PhoneNumber}, Адрес: {Address}");
    }

    public void UpdateContactInfo(string email, string phone, string address)
    {
        Email = email;
        PhoneNumber = phone;
        Address = address;
        Console.WriteLine("Контактная информация обновлена");
    }

    public string GetAccountStatus()
    {
        string status = IsActive ? "Активный" : "Неактивный";
        return $"Статус счета {AccountNumber}: {status}, Баланс: {Balance} {Currency}";
    }

    protected virtual void OnBalanceChangedEvent(decimal newBalance, decimal oldBalance)
    {
        OnBalanceChanged?.Invoke(newBalance, oldBalance);
    }

    public virtual void Deposit(decimal amount)
    {
        Deposit(amount, "Обычное пополнение");
    }

    public virtual void Deposit(decimal amount, string description)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма должна быть положительной");
        
        decimal oldBalance = Balance;
        Balance += amount;
        TotalTurnover += amount;
        TransactionsCount++;

        OnBalanceChangedEvent(Balance, oldBalance);
        
        Console.WriteLine($"{description}: +{amount} {Currency}. Новый баланс: {Balance} {Currency}");
    }

    public virtual void Withdraw(decimal amount)
    {
        Withdraw(amount, "Обычное снятие");
    }

    public virtual void Withdraw(decimal amount, string description)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма должна быть положительной");
            
        if (amount > Balance)
        {
            Console.WriteLine("Недостаточно средств");
            return;
        }

        decimal oldBalance = Balance;
        Balance -= amount;
        TotalTurnover += amount;
        TransactionsCount++;

        OnBalanceChangedEvent(Balance, oldBalance);
        
        Console.WriteLine($"{description}: -{amount} {Currency}. Новый баланс: {Balance} {Currency}");
    }

    public virtual void CalculateBenefits()
    {
        Console.WriteLine("Базовый счет не предоставляет специальных benefits");
    }

    public virtual void CalculateBenefits(int months)
    {
        Console.WriteLine($"Расчет benefits за {months} месяцев: 0 {Currency}");
    }

    public void SetDailyLimit(decimal limit)
    {
        DailyLimit = limit;
        Console.WriteLine($"Установлен дневной лимит: {limit} {Currency}");
    }

    public decimal CalculateAverageTransaction()
    {
        return TransactionsCount > 0 ? TotalTurnover / TransactionsCount : 0;
    }

    public void UpdateActiveDays()
    {
        ActiveDays++;
        Console.WriteLine($"Ваш счет активен уже {ActiveDays} дней");
    }

    public virtual void CloseAccount()
    {
        IsActive = false;
        Console.WriteLine($"Счет {AccountNumber} закрыт");
    }

    public virtual void CloseAccount(string reason)
    {
        IsActive = false;
        Console.WriteLine($"Счет {AccountNumber} закрыт. Причина: {reason}");
    }

    public virtual void FreezeAccount()
    {
        IsActive = false;
        Console.WriteLine($"Счет {AccountNumber} заморожен");
    }

    public virtual void UnfreezeAccount()
    {
        IsActive = true;
        Console.WriteLine($"Счет {AccountNumber} разморожен");
    }

    public virtual void ChangeCurrency(string newCurrency)
    {
        Currency = newCurrency;
        Console.WriteLine($"Валюта счета изменена на {newCurrency}");
    }
}

public class SavingsAccount : BankAccount, IAccountNotifier, IBalanceCalculator
{
    public decimal InterestRate { get; protected set; }
    public int DepositsCount { get; private set; }
    public decimal MaxBalance { get; private set; }
    public decimal MinBalanceRequirement { get; private set; }
    public int InterestAccrualPeriod { get; private set; }
    public decimal TotalInterestEarned { get; private set; }
    public DateTime LastInterestDate { get; private set; }
    public decimal YearlyInterest { get; private set; }
    public bool IsTaxFree { get; protected set; }
    public decimal CapitalizationAmount { get; private set; }
    public decimal EmergencyFund { get; private set; }
    public bool AutoRenewal { get; set; }
    public string SavingsGoal { get; set; }

    public decimal MonthlyDepositGoal { get; set; }
    public bool IsGoalAchieved { get; private set; }
    public DateTime GoalDeadline { get; set; }

    public SavingsAccount(string accountNumber, decimal initialBalance, decimal interestRate) 
        : base(accountNumber, initialBalance, "Сберегательный")
    {
        InterestRate = interestRate;
        MinBalanceRequirement = 1000;
        InterestAccrualPeriod = 30;
        LastInterestDate = DateTime.Now;
        MaxBalance = 1000000;
        DepositsCount = 0;
        YearlyInterest = 0;
        IsTaxFree = true;
        CapitalizationAmount = 0;
        MonthlyDepositGoal = 0;
        IsGoalAchieved = false;
        GoalDeadline = DateTime.Now.AddYears(1);
    }

    public void SetMonthlyDepositGoal(decimal goal)
    {
        MonthlyDepositGoal = goal;
        Console.WriteLine($"Цель ежемесячного пополнения установлена: {goal} {Currency}");
    }

    public void CheckGoalAchievement()
    {
        if (MonthlyDepositGoal > 0 && Balance >= MonthlyDepositGoal)
        {
            IsGoalAchieved = true;
            Console.WriteLine($"Поздравляем! Цель сбережений достигнута! Текущий баланс: {Balance} {Currency}");
        }
        else
        {
            Console.WriteLine($"Цель еще не достигнута. Текущий баланс: {Balance} {Currency}, Цель: {MonthlyDepositGoal} {Currency}");
        }
    }

    public void CalculateTimeToGoal()
    {
        if (MonthlyDepositGoal <= Balance)
        {
            Console.WriteLine("Цель уже достигнута!");
            return;
        }

        decimal needed = MonthlyDepositGoal - Balance;
        decimal monthlyInterest = Balance * InterestRate / 100;
        decimal monthlyGrowth = MonthlyDepositGoal > 0 ? MonthlyDepositGoal + monthlyInterest : monthlyInterest;
        
        if (monthlyGrowth <= 0)
        {
            Console.WriteLine("Недостаточный рост для достижения цели");
            return;
        }

        int months = (int)Math.Ceiling(needed / monthlyGrowth);
        Console.WriteLine($"Для достижения цели потребуется примерно {months} месяцев");
    }

    public override void Deposit(decimal amount)
    {
        base.Deposit(amount, "Пополнение сберегательного счета");
        DepositsCount++;

        if (Balance >= MinBalanceRequirement)
        {
            AccrueInterest();
        }

        CheckGoalAchievement();
    }

    public override void Deposit(decimal amount, string description)
    {
        base.Deposit(amount, description);
        DepositsCount++;
        CapitalizationAmount += amount;

        CheckGoalAchievement();
    }

    public override void CalculateBenefits()
    {
        decimal monthlyInterest = Balance * InterestRate / 100;
        Console.WriteLine($"Ежемесячные проценты: {monthlyInterest} {Currency}");
    }

    public override void CalculateBenefits(int months)
    {
        decimal totalInterest = Balance * InterestRate / 100 * months;
        Console.WriteLine($"Проценты за {months} месяцев: {totalInterest} {Currency}");
    }

    public void SetMinBalanceRequirement(decimal amount)
    {
        MinBalanceRequirement = amount;
        Console.WriteLine($"Минимальное требование к балансу: {amount} {Currency}");
    }

    public void AccrueInterest()
    {
        if (Balance >= MinBalanceRequirement)
        {
            decimal interest = Balance * InterestRate / 100;
            TotalInterestEarned += interest;
            YearlyInterest += interest;
            LastInterestDate = DateTime.Now;
            
            decimal oldBalance = Balance;
            Balance += interest;

            OnBalanceChangedEvent(Balance, oldBalance);
            
            Console.WriteLine($"Начислены проценты: {interest} {Currency}");
        }
    }

    public decimal GetProjectedInterest(int days)
    {
        return Balance * InterestRate / 100 * (days / (decimal)InterestAccrualPeriod);
    }

    public string GetInterestHistory()
    {
        return $"Заработано процентов: {TotalInterestEarned} {Currency}";
    }

    public void ApplyTaxDeduction()
    {
        if (IsTaxFree && YearlyInterest > 0)
        {
            Console.WriteLine($"Применена налоговая льгота на проценты: {YearlyInterest} {Currency}");
            YearlyInterest = 0;
        }
    }

    public void SetEmergencyFund(decimal amount)
    {
        EmergencyFund = amount;
        Console.WriteLine($"Установлен фонд экстренных сбережений: {amount} {Currency}");
    }

    public void SetSavingsGoal(string goal)
    {
        SavingsGoal = goal;
        Console.WriteLine($"Цель сбережений установлена: {goal}");
    }

    public void ToggleAutoRenewal()
    {
        AutoRenewal = !AutoRenewal;
        Console.WriteLine($"Автопродление: {(AutoRenewal ? "включено" : "выключено")}");
    }

    public void SendNotification(string message)
    {
        Console.WriteLine($"Уведомление для сберегательного счета {AccountNumber}: {message}");
    }
    
    public decimal CalculateProjectedBalance(int months)
    {
        decimal projectedBalance = Balance;
        for (int i = 0; i < months; i++)
        {
            projectedBalance += projectedBalance * InterestRate / 100;
        }
        return projectedBalance;
    }
}

public class CheckingAccount : BankAccount, IAccountNotifier, IBalanceCalculator
{
    public decimal OverdraftLimit { get; private set; }
    public decimal MonthlyFee { get; private set; }
    public bool IsVip { get; private set; }
    public int FreeTransactionsLeft { get; private set; }
    public decimal TransactionCommission { get; private set; }
    public DateTime NextFeeDate { get; private set; }
    public int OverdraftUsageCount { get; private set; }
    public decimal UsedOverdraft { get; private set; }
    public decimal CashbackRate { get; private set; }
    public decimal MonthlyCashback { get; private set; }
    public int CardLevel { get; private set; }
    public decimal DailyWithdrawalLimit { get; set; }
    public bool OverdraftProtection { get; set; }
    public string LinkedCardNumber { get; set; }

    public int MonthlyTransactionLimit { get; set; }
    public bool IsInternational { get; set; }
    public string CardType { get; set; }

    public CheckingAccount(string accountNumber, decimal initialBalance, decimal overdraftLimit) 
        : base(accountNumber, initialBalance, "Текущий")
    {
        OverdraftLimit = overdraftLimit;
        MonthlyFee = 50;
        FreeTransactionsLeft = 10;
        TransactionCommission = 10;
        NextFeeDate = DateTime.Now.AddMonths(1);
        IsVip = false;
        UsedOverdraft = 0;
        CashbackRate = 1.0m;
        MonthlyCashback = 0;
        CardLevel = 1;

        MonthlyTransactionLimit = 50;
        IsInternational = false;
        CardType = "Стандартная";
    }

    public void SetMonthlyTransactionLimit(int limit)
    {
        MonthlyTransactionLimit = limit;
        Console.WriteLine($"Лимит месячных транзакций установлен: {limit}");
    }

    public void ToggleInternationalStatus()
    {
        IsInternational = !IsInternational;
        Console.WriteLine($"Международный статус: {(IsInternational ? "включен" : "выключен")}");
    }

    public void ChangeCardType(string newCardType)
    {
        CardType = newCardType;
        Console.WriteLine($"Тип карты изменен на: {newCardType}");
    }

    public override void Withdraw(decimal amount)
    {
        Withdraw(amount, "Снятие с текущего счета");
    }

    public override void Withdraw(decimal amount, string description)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма должна быть положительной");

        if (TransactionsCount >= MonthlyTransactionLimit)
        {
            Console.WriteLine($"Достигнут месячный лимит транзакций: {MonthlyTransactionLimit}");
            return;
        }
        
        decimal availableAmount = Balance + OverdraftLimit;
        if (amount > availableAmount)
        {
            Console.WriteLine($"Недостаточно средств. Доступно: {availableAmount} {Currency}");
            return;
        }

        decimal oldBalance = Balance;

        if (amount > Balance)
        {
            decimal overdraftUsed = amount - Balance;
            UsedOverdraft = overdraftUsed;
            OverdraftUsageCount++;
            Console.WriteLine($"Использован овердрафт на сумму {overdraftUsed} {Currency}");
        }
        
        Balance -= amount;
        TotalTurnover += amount;
        TransactionsCount++;

        decimal cashback = amount * CashbackRate / 100;
        MonthlyCashback += cashback;

        if (FreeTransactionsLeft > 0)
            FreeTransactionsLeft--;
        else
        {
            Balance -= TransactionCommission;
            Console.WriteLine($"Списана комиссия за операцию: {TransactionCommission} {Currency}");
        }

        OnBalanceChangedEvent(Balance, oldBalance);
            
        Console.WriteLine($"{description}: -{amount} {Currency}. Новый баланс: {Balance} {Currency}");
        Console.WriteLine($"Начислен кэшбэк: {cashback} {Currency}");
        Console.WriteLine($"Использовано овердрафта: {Math.Max(0, -Balance)}/{OverdraftLimit} {Currency}");
    }

    public override void CalculateBenefits()
    {
        Console.WriteLine($"Текущий кэшбэк: {MonthlyCashback} {Currency}, Ставка: {CashbackRate}%");
    }

    public override void CalculateBenefits(int months)
    {
        decimal projectedCashback = MonthlyCashback * months;
        Console.WriteLine($"Расчет кэшбэка за {months} месяцев: {projectedCashback} {Currency}");
    }

    public void SetTransactionCommission(decimal commission)
    {
        TransactionCommission = commission;
        Console.WriteLine($"Комиссия: {commission} {Currency}");
    }

    public decimal CalculateTotalMonthlyCosts()
    {
        return MonthlyFee + (TransactionsCount * TransactionCommission);
    }

    public string GetOverdraftStatus()
    {
        decimal usedOverdraft = Balance < 0 ? Math.Abs(Balance) : 0;
        string status = usedOverdraft > 0 ? "Используется" : "Не используется";
        return $"Статус овердрафта: {status} ({usedOverdraft}/{OverdraftLimit} {Currency})";
    }

    public void UpdateFeeDate(DateTime newDate)
    {
        NextFeeDate = newDate;
        Console.WriteLine($"Обновлена дата следующей комиссии: {newDate:d}");
    }

    public void RepayOverdraft(decimal amount)
    {
        if (Balance >= 0)
        {
            Console.WriteLine("Овердрафт не используется");
            return;
        }

        decimal overdraftAmount = Math.Abs(Balance);
        decimal repayment = Math.Min(amount, overdraftAmount);
        
        decimal oldBalance = Balance;
        Balance += repayment;

        OnBalanceChangedEvent(Balance, oldBalance);
        
        Console.WriteLine($"Погашено овердрафта: {repayment} {Currency}");
        Console.WriteLine($"Остаток овердрафта: {Math.Max(0, -Balance)} {Currency}");
    }

    public void ApplyMonthlyCashback()
    {
        if (MonthlyCashback > 0)
        {
            decimal oldBalance = Balance;
            Balance += MonthlyCashback;

            OnBalanceChangedEvent(Balance, oldBalance);
            
            Console.WriteLine($"Начислен месячный кэшбэк: {MonthlyCashback} {Currency}");
            MonthlyCashback = 0;
        }
    }

    public void SetDailyWithdrawalLimit(decimal limit)
    {
        DailyWithdrawalLimit = limit;
        Console.WriteLine($"Установлен дневной лимит снятия: {limit} {Currency}");
    }

    public void ToggleOverdraftProtection()
    {
        OverdraftProtection = !OverdraftProtection;
        Console.WriteLine($"Защита от овердрафта: {(OverdraftProtection ? "включена" : "выключена")}");
    }

    public void LinkCard(string cardNumber)
    {
        LinkedCardNumber = cardNumber;
        Console.WriteLine($"Карта {cardNumber} привязана к счету");
    }

    public void SendNotification(string message)
    {
        Console.WriteLine($"Уведомление для текущего счета {AccountNumber}: {message}");
    }

    public decimal CalculateProjectedBalance(int months)
    {
        return Balance - (MonthlyFee * months);
    }
}

public class PremiumAccount : BankAccount, ITransactionLogger, IAuditService
{
    public decimal InterestRate { get; protected set; }
    public decimal PremiumBonusRate { get; private set; }
    public int LoyaltyProgramPoints { get; private set; }
    public DateTime PremiumExpirationDate { get; private set; }
    public decimal TotalPremiumBenefits { get; private set; }
    public decimal TotalInterestEarned { get; private set; }
    public DateTime LastInterestDate { get; private set; }

    public string PremiumLevel { get; set; }
    public bool HasConciergeService { get; set; }
    public decimal InsuranceCoverage { get; set; }

    public PremiumAccount(string accountNumber, decimal initialBalance, decimal interestRate) 
        : base(accountNumber, initialBalance, "Премиум")
    {
        InterestRate = interestRate;
        PremiumBonusRate = 2.0m;
        LoyaltyProgramPoints = 1000;
        PremiumExpirationDate = DateTime.Now.AddYears(1);
        TotalPremiumBenefits = 0;
        TotalInterestEarned = 0;
        LastInterestDate = DateTime.Now;

        PremiumLevel = "Gold";
        HasConciergeService = true;
        InsuranceCoverage = 100000;
    }

    public void ChangePremiumLevel(string newLevel)
    {
        PremiumLevel = newLevel;
        Console.WriteLine($"Уровень премиум изменен на: {newLevel}");
    }

    public void ToggleConciergeService()
    {
        HasConciergeService = !HasConciergeService;
        Console.WriteLine($"Услуга консьержа: {(HasConciergeService ? "включена" : "выключена")}");
    }

    public void SetInsuranceCoverage(decimal coverage)
    {
        InsuranceCoverage = coverage;
        Console.WriteLine($"Страховое покрытие установлено: {coverage} {Currency}");
    }

    public void LogTransaction(string message)
    {
        Console.WriteLine($"Банковская выписка счета {AccountNumber}: {message} - {DateTime.Now}");
    }

    public void AuditTransaction(decimal amount)
    {
        Console.WriteLine($"Аудиторская проверка счета {AccountNumber}: Сумма {amount}, Баланс {Balance}, Время {DateTime.Now}");
    }

    public override void Deposit(decimal amount)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма должна быть положительной");
        
        decimal oldBalance = Balance;
        Balance += amount;
        TotalTurnover += amount;
        TransactionsCount++;
        decimal bonus = CalculatePremiumBonus(amount);
        Balance += bonus;
        TotalPremiumBenefits += bonus;
     
        AddLoyaltyPoints((int)(amount / 100));
        AccrueInterest();
        LogTransaction($"Депозит: {amount} руб; Полученный бонус: {bonus} руб.");
        AuditTransaction(amount);

        OnBalanceChangedEvent(Balance, oldBalance);
        
        Console.WriteLine($"Новый баланс: {Balance} руб.");
    }

    public override void Withdraw(decimal amount)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма должна быть положительной");
            
        if (amount > Balance)
        {
            Console.WriteLine("Недостаточно средств");
            return;
        }

        decimal oldBalance = Balance;
        Balance -= amount;
        TotalTurnover += amount;
        TransactionsCount++;
        
        LogTransaction($"Снятие: {amount} руб.");
        AuditTransaction(amount);

        OnBalanceChangedEvent(Balance, oldBalance);
        
        Console.WriteLine($"Новый баланс: {Balance} руб.");
    }

    public void AddLoyaltyPoints(int points)
    {
        LoyaltyProgramPoints += points;
        Console.WriteLine($"Начислено баллов лояльности: {points}");
    }

    public decimal CalculatePremiumBonus(decimal amount)
    {
        return amount * PremiumBonusRate / 100;
    }

    private void AccrueInterest()
    {
        decimal interest = Balance * InterestRate / 100;
        TotalInterestEarned += interest;
        LastInterestDate = DateTime.Now;
        decimal oldBalance = Balance;
        Balance += interest;

        OnBalanceChangedEvent(Balance, oldBalance);
        
        Console.WriteLine($"Начислены премиум проценты: {interest} руб.");
    }

    public string GetPremiumStatus()
    {
        return $"Премиум статус действителен до: {PremiumExpirationDate:d}, Уровень: {PremiumLevel}";
    }

    public string GetInterestHistory()
    {
        return $"Заработано процентов: {TotalInterestEarned} руб., Премиум бонусов: {TotalPremiumBenefits} руб.";
    }
}

public class AccountManager
{
    private readonly IAccountNotifier _notifier;
    private readonly IBalanceCalculator _calculator;

    public AccountManager(IAccountNotifier notifier, IBalanceCalculator calculator)
    {
        _notifier = notifier;
        _calculator = calculator;
    }

    public void PerformAccountOperations(string accountNumber)
    {
        _notifier.SendNotification($"Операции выполняются для счета {accountNumber}");
        
        decimal projectedBalance = _calculator.CalculateProjectedBalance(12);
        Console.WriteLine($"Прогнозируемый баланс через 12 месяцев: {projectedBalance} {GetCurrency(accountNumber)}");
    }

    private string GetCurrency(string accountNumber)
    {
        return "RUB";
    }
}


public class AccountEventHandler
{
    public static void HandleBalanceChange(decimal newBalance, decimal oldBalance)
    {
        Console.WriteLine($"Баланс изменился: {oldBalance} -> {newBalance}. Разница: {newBalance - oldBalance}");
    }

    public static void HandleCollectionChange(string message)
    {
        Console.WriteLine($"Событие коллекции: {message}");
    }
}


BankAccountCollection<BankAccount> bank = new BankAccountCollection<BankAccount>();
bank.OnCollectionChanged += AccountEventHandler.HandleCollectionChange;

SavingsAccount savings = new SavingsAccount("S00001", 5000, 5) { OwnerName = "Иван Иванов" };
CheckingAccount checking = new CheckingAccount("C00001", 2000, 1000) { OwnerName = "Петр Петров" };
PremiumAccount premium = new PremiumAccount("P00001", 10000, 3) { OwnerName = "Мария Сидорова" };
        
savings.OnBalanceChanged += AccountEventHandler.HandleBalanceChange;
checking.OnBalanceChanged += AccountEventHandler.HandleBalanceChange;
premium.OnBalanceChanged += AccountEventHandler.HandleBalanceChange;

bank.Add(savings);
bank.Add(checking);
bank.Add(premium);
                        
List<BankAccount> accounts = new List<BankAccount> { savings, checking, premium };

Console.WriteLine($"Всего счетов в коллекции: {bank.GetTotalAccounts()}");
Console.WriteLine($"Общий баланс всех счетов: {bank.GetTotalBalance()}");


foreach (var account in accounts)
{

    Console.WriteLine(account.GetInfo(true));

    account.DisplayContactInfo();
    account.UpdateContactInfo("test@email.com", "+79991234567", "Москва");
    Console.WriteLine(account.GetAccountStatus());
            
    account.Deposit(1000);
    account.Deposit(500, "Специальное пополнение");
    account.CalculateBenefits();
    account.CalculateBenefits(6);
    account.UpdateActiveDays();
}

savings.SetMonthlyDepositGoal(8000);
savings.CheckGoalAchievement();
savings.CalculateTimeToGoal();

checking.SetMonthlyTransactionLimit(30);
checking.ToggleInternationalStatus();
checking.ChangeCardType("Платиновая");

premium.ChangePremiumLevel("Platinum");
premium.ToggleConciergeService();
premium.SetInsuranceCoverage(200000);

savings.AccrueInterest();
savings.ApplyTaxDeduction();
Console.WriteLine(savings.GetInterestHistory());

checking.Withdraw(2500);
Console.WriteLine(checking.GetOverdraftStatus());
checking.ApplyMonthlyCashback();

premium.AddLoyaltyPoints(500);
Console.WriteLine(premium.GetPremiumStatus());

bank.DisplayAllAccounts();

IAccountNotifier savingsNotifier = (IAccountNotifier)savings;
savingsNotifier.SendNotification("Тестовое уведомление для сберегательного счета");

IBalanceCalculator savingsCalculator = (IBalanceCalculator)savings;
decimal projectedSavings = savingsCalculator.CalculateProjectedBalance(12);
Console.WriteLine($"Прогноз баланса сберегательного счета: {projectedSavings}");

IAccountNotifier checkingNotifier = (IAccountNotifier)checking;
checkingNotifier.SendNotification("Тестовое уведомление для текущего счета");

IBalanceCalculator checkingCalculator = (IBalanceCalculator)checking;
decimal projectedChecking = checkingCalculator.CalculateProjectedBalance(12);
Console.WriteLine($"Прогноз баланса текущего счета: {projectedChecking}");

AccountManager savingsManager = new AccountManager(savingsNotifier, savingsCalculator);
savingsManager.PerformAccountOperations("S00001");

AccountManager checkingManager = new AccountManager(checkingNotifier, checkingCalculator);
checkingManager.PerformAccountOperations("C00001");

savings.ChangeCurrency("USD");
checking.FreezeAccount();
checking.UnfreezeAccount();

savings.SetEmergencyFund(2000);
savings.SetSavingsGoal("Покупка квартиры");
savings.ToggleAutoRenewal();

checking.SetDailyWithdrawalLimit(30000);
checking.ToggleOverdraftProtection();
checking.LinkCard("1234-5678-9012-3456");

```
